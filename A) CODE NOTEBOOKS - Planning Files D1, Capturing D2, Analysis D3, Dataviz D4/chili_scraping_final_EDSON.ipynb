```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from IPython.display import Image, display
```

```python
def scrape_chili(name, url):
    headers = {"User-Agent": "Mozilla/5.0"}
    
    response = requests.get(url, headers=headers)
    
    if response.status_code != 200:
        return {"name": name, "image": None}
    
    soup = BeautifulSoup(response.text, "html.parser")
    infobox = soup.find("table", class_="infobox")
    
    img_url = None
    if infobox:
        img = infobox.find("img")
        if img:
            img_url = "https:" + img["src"]
    
    return {
        "name": name,
        "image": img_url
    }
```

```python
missing_chiles = [
    ("Habanero / Scotch Bonnet", "https://en.wikipedia.org/wiki/Habanero"),
    ("Paprika / Bulgarian-type", "https://en.wikipedia.org/wiki/Bell_pepper"),
    ("Espelette / Basque pepper", "https://en.wikipedia.org/wiki/Piment_d%27Espelette"),
    ("Paprika (Hungarian-type)", "https://en.wikipedia.org/wiki/Paprika"),
    ("Shatta", "https://en.wikipedia.org/wiki/Shatta"),
    ("Italian peperoncino", "https://en.wikipedia.org/wiki/Peperoncino"),
    ("Small hot peperoncini", "https://en.wikipedia.org/wiki/Peperoncini"),
    ("Harissa-specific varieties", "https://en.wikipedia.org/wiki/Harissa"),
    ("Piri Piri", "https://en.wikipedia.org/wiki/Piri_piri"),
    ("Padron", "https://en.wikipedia.org/wiki/Padr%C3%B3n_pepper"),
    ("Espelette / Aleppo-like", "https://en.wikipedia.org/wiki/Aleppo_pepper"),
    ("Scotch Bonnet (Caribbean market)", "https://en.wikipedia.org/wiki/Scotch_bonnet")
]
```

```python
new_chiles = []

for name, url in missing_chiles:
    print("Procesando:", name)
    new_chiles.append(scrape_chili(name, url))
```

```python
df = pd.DataFrame(new_chiles)
df
```

```python
for chile in new_chiles:
    print(chile["name"])
    if chile["image"]:
        display(Image(url=chile["image"]))
```

```python
df.to_csv("chiles_final.csv", index=False, encoding="utf-8")
print("Archivo guardado correctamente")
```
